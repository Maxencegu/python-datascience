# **Introduction**

Au TD précédent, vous avez appris à stocker des données dans des variables et des listes,
et à les manipuler en Python. Vous savez maintenant accéder à un élément précis,
extraire une tranche, modifier ou supprimer des valeurs.

Mais imaginez que vous devez calculer l'IMC de 500 étudiants.
Réécrire la même formule 500 fois serait laborieux — et une seule erreur de frappe
invaliderait le résultat. Et si vous vouliez ensuite calculer une moyenne,
un écart-type ou une corrélation sur ces 500 valeurs ?
Les listes Python ne sont pas conçues pour ce genre de calcul.

C'est là qu'interviennent les **fonctions** et **NumPy**.

Une fonction encapsule une opération réutilisable sous un nom clair.
Python en fournit de nombreuses d'emblée — `len()`, `max()`, `round()` — et vous pouvez
en créer vous-même pour structurer vos propres analyses.
Un **module** est une bibliothèque de fonctions prête à l'emploi : il suffit de l'importer
pour étendre les capacités de Python sans réécrire ce qui existe déjà.

**NumPy** (*Numerical Python*) est le module de référence pour le calcul scientifique.
Il introduit un nouveau type de données — le **tableau NumPy** — qui permet d'effectuer
des opérations mathématiques sur des milliers de valeurs en une seule ligne de code,
bien plus efficacement qu'avec une liste Python classique.

Dans ce TD, vous apprendrez à :
- utiliser les fonctions intégrées de Python (`len`, `max`, `min`, `round`...) ;
- créer vos propres fonctions pour rendre votre code réutilisable ;
- importer un module et accéder à ses fonctions ;
- créer des tableaux NumPy et comprendre leur avantage sur les listes ;
- effectuer des opérations arithmétiques sur des tableaux entiers en une ligne ;
- calculer des statistiques descriptives (moyenne, écart-type, médiane) avec NumPy.

À l'issue de cette séance, vous serez capables d'analyser un jeu de données numérique
complet sans boucle, sans répétition — et en quelques lignes de code.

```python

plt.scatter(x, y)
```

Dans l'exercice précédent, vous n'avez pas spécifié le nombre de classes. Python
fixe alors ce nombre à 10 par défaut.

Le choix du nombre de classes est important :

### **Exercice**  
**Effets secondaires de NumPy**

**NumPy** est très pratique pour effectuer des opérations **vectorielles**. Cependant, si vous comparez son fonctionnement avec celui des **listes Python classiques**, certaines différences apparaissent.

Premièrement, les **tableaux NumPy** ne peuvent pas contenir des éléments de types différents. Si vous mélangez des types, par exemple des **booléens** et des **entiers**, NumPy les convertit automatiquement vers un **type commun**.  
Les booléens comme `True` et `False` sont alors interprétés comme `1` et `0` lorsqu’ils sont combinés avec des nombres, ce qui fait que le tableau devient un tableau d’**entiers**.

Deuxièmement, les opérateurs arithmétiques habituels comme `+`, `-`, `*` et `/` n’ont **pas la même signification** pour les listes Python classiques et pour les tableaux NumPy.

### **Exercice**
**Sous-sélection de tableaux NumPy 2D**

Si votre **tableau NumPy 2D** a une structure régulière, c’est-à-dire que chaque ligne et chaque colonne contient un nombre fixe de valeurs, les **sous-sélections compliquées deviennent très simples**.  
Regardez l’exemple ci-dessous où les éléments `"a"` et `"c"` sont extraits d’une liste de listes :

```python
# numpy
import numpy as np
np_x = np.array(x)
np_x[:, 0]
```
Les indices avant la virgule font référence aux **lignes**, tandis que ceux après la virgule font référence aux **colonnes**.  
Le symbole `:` sert à **slicer** ; dans cet exemple, il indique à Python d’inclure **toutes les lignes**.
