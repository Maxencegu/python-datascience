Sélectionner des valeurs individuelles dans une liste n’est qu’une partie du travail.  
Il est également possible de **découper une liste** (*slicing*), c’est-à-dire sélectionner **plusieurs éléments à la fois**.

Utilisez la syntaxe suivante :

```python

```
L’index de début (**start**) est **inclus**, tandis que l’index de fin (**end**) ne l’est pas.  
Il est également possible de ne pas préciser ces index.  
Si vous n’indiquez pas l’index de début, Python comprend que vous souhaitez commencer la découpe au **début de la liste**.

### Points clés abordés

- **Indexation (Indexing)**  
  Vous avez accédé aux éléments d’une liste grâce à leur index.  
  Exemple :
  ```python
  fam[3]
  ```
  → récupère le **quatrième élément**.

- **Indexation négative (Negative indexing)**  
  Vous avez utilisé des index négatifs pour compter à rebours depuis la fin de la liste.  
  Exemple :
  ```python
  fam[-1]
  ```
  → récupère le **dernier élément**.

### Exemples clés

- **Indexation de base (Basic Indexing)**  
  Accéder à un élément précis d’une liste à l’aide de son index.

  ```python
  x = ["a", "b", "c", "d"]
  print(x[1])  # Output: "b"
  print(x[-3])  # Output: "b"
  ```
- **Slicing**

  ```python
  x = ["a", "b", "c", "d"]
  print(x[1:3])  # Output: ["b", "c"]
  ```

  Exemple : ajouter votre nom et votre taille à la liste `fam` :
  ```python
  fam_ext = fam + ["your_name", 1.80]
  ```
- **Supprimer des éléments (Deleting Elements)**
  L’instruction `del` permet de supprimer un élément de la liste.
  Exemple : supprimer l’élément à l’index 2 :
  ```python
  del fam[2]
  ```
- **Copier des listes (Copying Lists)**
  Utiliser simplement `=` ne crée pas une vraie copie mais seulement une référence à la liste.
  Pour créer une copie indépendante, utilisez `list()` ou le slicing :
  ```python
  y = list(x)  # ou x[:]
  ```

Ces notions sont fondamentales pour travailler efficacement avec les données en Python.
