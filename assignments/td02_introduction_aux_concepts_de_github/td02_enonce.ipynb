# **Introduction**

Au TD précédent, vous avez appris à versionner votre travail localement avec Git :
créer des commits, consulter l'historique, comparer des versions, annuler une erreur.
Vos modifications étaient sauvegardées — mais uniquement sur votre propre machine.

Imaginez maintenant que vous travaillez à deux sur un modèle de prévision économique.
L'un d'entre vous ajoute des données, l'autre modifie le script.
Comment éviter d'écraser le travail de l'autre ?
Comment savoir qui a fait quoi, et pourquoi ?
Comment partager votre code avec votre directeur de mémoire sans envoyer des fichiers par e-mail ?

C'est exactement ce que **GitHub** résout.

GitHub est un service d'hébergement dans le cloud qui s'appuie sur Git pour rendre votre travail
**accessible, collaboratif et traçable**.
Chaque dépôt que vous créez sur GitHub est une copie distante de votre projet :
vous pouvez le partager avec un collègue, recevoir des retours via une Pull Request,
ou déclencher automatiquement des vérifications à chaque modification.

Dans ce TD, vous apprendrez à :
- créer et configurer un dépôt GitHub pour vos projets de données ;
- rédiger un fichier README clair pour présenter votre travail ;
- déposer et organiser vos notebooks sur GitHub ;
- utiliser les branches pour travailler sans risque sur votre code ;
- collaborer via les Issues et les Pull Requests ;
- mettre en place les tests automatiques qui vérifieront votre travail à chaque TD.

À l'issue de cette séance, votre dépôt personnel `upjv-python-datascience` sera
configuré et prêt pour toute l'année.

```python
# @title ⚙️ Préparation de l'exercice (exécuter d'abord)
import requests

try:
    del tout_configure
except NameError:
    pass

REPO = "upjv-python-datascience"

print("État actuel de votre dépôt :\n")
try:
    checks = {
        "Branche dev_td03": False,
        "Workflow GitHub Actions": False,
        "Protection de main": False,
    }

    # Branches
    r = requests.get(f"https://api.github.com/repos/{github_username}/{REPO}/branches")
    if r.status_code == 200:
        branches = [b["name"] for b in r.json()]
        checks["Branche dev_td03"] = "dev_td03" in branches

    # Workflow
    r = requests.get(
        f"https://api.github.com/repos/{github_username}/{REPO}/contents/"
        f".github/workflows/tests_td.yml"
    )
    checks["Workflow GitHub Actions"] = r.status_code == 200

    # Protection main
    r = requests.get(
        f"https://api.github.com/repos/{github_username}/{REPO}/branches/main"
    )
    if r.status_code == 200:
        checks["Protection de main"] = r.json().get("protected", False)

    for label, ok in checks.items():
        print(f"  {'✅' if ok else '⏳'} {label}")

except NameError:
    print("  ❌ github_username non défini — exécutez d'abord l'exercice précédent")
```

```python
# @title 🔧 Exercice — Sauvegarder le TD02 via une Pull Request
# ================================================================
#  EXERCICE — Premier cycle complet dev_td02 → PR → merge [ 6 étapes ]
# ================================================================
#
# Référez-vous aux infographies 11 (Pull Requests) et 12 (Examiner une PR)
#
# ----------------------------------------------------------------
# Étape 1/6 — Créer la branche dev_td02
# ----------------------------------------------------------------
# Sur GitHub : onglet Code → icône branches → New branch
# Nom : dev_td02  |  Source : main
#
# ----------------------------------------------------------------
# Étape 2/6 — Uploader votre notebook sur dev_td02
# ----------------------------------------------------------------
# Téléchargez votre notebook depuis Colab : Fichier → Télécharger → .ipynb
# Sur GitHub : sélectionner la branche dev_td02 → Add file → Upload files
# Fichier à uploader : td02_enonce.ipynb
#
# ----------------------------------------------------------------
# Étape 3/6 — Ouvrir une Pull Request
# ----------------------------------------------------------------
# Onglet Pull requests → New pull request
# compare : dev_td02  →  base : main
# Titre : TD02 — Notebook complété
#
# ----------------------------------------------------------------
# Étape 4/6 — Désigner votre collègue comme reviewer
# ----------------------------------------------------------------
# Section Reviewers (icône engrenage) → sélectionner votre collègue
#
# ----------------------------------------------------------------
# Étape 5/6 — Votre collègue approuve la PR si le travail est bien fait
# ----------------------------------------------------------------
# Côté reviewer : onglet Files changed → Review changes → Approve → Submit review
#
# ----------------------------------------------------------------
# Étape 6/6 — Merger et supprimer la branche
# ----------------------------------------------------------------
# Merge pull request → Confirm merge → Delete branch
```

```python
# @title ✅ Vérification
import requests

REPO = "upjv-python-datascience"
print("Résultats :\n")

r_main   = requests.get(f"https://api.github.com/repos/{github_username}/{REPO}/contents/td02_enonce.ipynb",
                        params={"ref": "main"})
r_branch = requests.get(f"https://api.github.com/repos/{github_username}/{REPO}/branches/dev_td02")

if r_main.status_code == 200:
    print("  ✅ td02_enonce.ipynb est présent sur main — PR mergée avec succès !")
elif r_branch.status_code == 200:
    r_file = requests.get(f"https://api.github.com/repos/{github_username}/{REPO}/contents/td02_enonce.ipynb",
                          params={"ref": "dev_td02"})
    if r_file.status_code == 200:
        print("  ⏳ Notebook uploadé sur dev_td02 — en attente d'approbation et de merge")
        print("      Faites approuver la PR par votre collègue puis mergez sur main")
    else:
        print("  ❌ Branche dev_td02 créée mais notebook non uploadé")
        print("      Téléchargez votre notebook Colab et uploadez-le sur la branche dev_td02")
else:
    print("  ❌ Branche dev_td02 introuvable — créez-la depuis main sur GitHub")
```
