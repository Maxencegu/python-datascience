### Python vs Java — même tâche, moins de code

````python
# Python : calculer une moyenne
notes = [12, 15, 18, 10, 14]
moyenne = sum(notes) / len(notes)
print(f"Moyenne : {moyenne}")
````

```java
// Java : même calcul — beaucoup plus verbeux
public class Moyenne {
    public static void main(String[] args) {
        int[] notes = {12, 15, 18, 10, 14};
        int somme = 0;
        for (int note : notes) { somme += note; }
        double moyenne = (double) somme / notes.length;
        System.out.println("Moyenne : " + moyenne);
    }
}
```

## 2. Qu'est-ce que la programmation ?

Programmer, c'est **écrire des instructions étape par étape pour qu'un ordinateur les exécute**. Comme une recette de cuisine : chaque étape doit être précise et dans le bon ordre.

Les ordinateurs ne comprennent que le binaire (0 et 1). Python fait le pont :

```
Vous écrivez du Python (lisible)  →  Python traduit  →  L'ordinateur exécute
```

### Votre premier aperçu de Python

````python
# Ce code est presque lisible comme de l'anglais
nom = "Alice"
age = 20
print(f"Bonjour, je m'appelle {nom} et j'ai {age} ans.")

if age >= 18:
    print("Je suis majeur.")
else:
    print("Je suis mineur.")
````

## 7. Environnements virtuels

### 7.1 Pourquoi un environnement virtuel ?

Imaginez deux projets sur votre ordinateur :
- **Projet A** : nécessite `pandas` version 1.5
- **Projet B** : nécessite `pandas` version 2.0

Sans environnement virtuel, impossible d'avoir les deux — une seule version peut être installée globalement.

**Un environnement virtuel = une installation Python isolée par projet.** Chaque projet a ses propres paquets, ses propres versions, complètement indépendants.

```
PythonProjects/
├── projet-a/
│   ├── .venv/          ← pandas 1.5 installé ici
│   └── script.py
└── projet-b/
    ├── .venv/          ← pandas 2.0 installé ici
    └── script.py
```

### 7.2 Créer un environnement virtuel

Dans le terminal VS Code (`` Ctrl+` ``)  :

```bash
python -m venv .venv
```

Cela crée un dossier `.venv` dans votre projet avec une installation Python isolée.

> **Bonne nouvelle** : VS Code détecte automatiquement le `.venv` et l'active dans le terminal intégré — vous verrez `(.venv)` devant votre invite de commande.

### 7.3 Activer manuellement

Si VS Code ne l'active pas automatiquement :

```bash
# Windows
.venv\Scripts\activate

# Mac/Linux
source .venv/bin/activate
```

Le `(.venv)` en début de ligne confirme que l'environnement est actif.

### 7.4 Désactiver

```bash
deactivate
```

### 7.5 Note sur Anaconda

Vous avez peut-être entendu parler d'Anaconda — un outil qui gère aussi les environnements Python. Il est populaire en data science car il inclut beaucoup de paquets par défaut.

**Recommandation** : utilisez les environnements virtuels Python natifs (`venv`). C'est plus léger, plus rapide, et c'est le standard en entreprise. N'utilisez Anaconda que si un projet ou une équipe le requiert spécifiquement.

## 8. Paquets Python et pip

### 8.1 Qu'est-ce qu'un paquet ?

Un paquet est du **code Python écrit par d'autres développeurs** que vous pouvez utiliser dans vos projets. Plutôt que de tout réécrire de zéro, vous réutilisez ce qui existe.

Exemples de paquets essentiels :

| Paquet | Utilisation |
|--------|------------|
| `requests` | Télécharger des données depuis le web |
| `pandas` | Manipuler des données tabulaires |
| `numpy` | Calculs mathématiques rapides |
| `matplotlib` | Créer des graphiques |
| `openai` | Connecter une application à ChatGPT |

### 8.2 pip — le gestionnaire de paquets

`pip` (Pip Installs Packages) télécharge les paquets depuis **PyPI** (Python Package Index) — une bibliothèque de plus de 500 000 paquets open-source.

`pip` est automatiquement inclus avec Python.

### 8.3 Commandes essentielles

```bash
# Installer un paquet
pip install requests

# Installer une version spécifique
pip install pandas==2.1.0

# Installer plusieurs paquets
pip install numpy matplotlib seaborn

# Lister les paquets installés
pip list

# Désinstaller un paquet
pip uninstall requests
```

### 8.4 Utiliser un paquet installé

````python
# Après pip install requests, on peut l'importer et l'utiliser
import requests

reponse = requests.get("https://api.github.com")
print(f"Statut de l'API GitHub : {reponse.status_code}")
# 200 = succès
````

### 8.5 Le fichier `requirements.txt`

Pour partager votre projet, listez tous vos paquets dans un fichier `requirements.txt` :

```bash
# Générer automatiquement la liste des paquets installés
pip freeze > requirements.txt
```

Contenu exemple de `requirements.txt` :
```
pandas==2.1.0
numpy==1.26.0
matplotlib==3.8.0
requests==2.31.0
```

Un collègue peut recréer l'environnement identique avec :
```bash
pip install -r requirements.txt
```

### 8.6 Méthodes d'import

````python
# Méthode 1 : importer le module entier
import math
print(math.pi)          # 3.14159...
print(math.sqrt(16))    # 4.0

# Méthode 2 : importer une fonction spécifique
from math import pi, sqrt
print(pi)               # 3.14159...
print(sqrt(25))         # 5.0

# Méthode 3 : importer avec un alias (convention pour pandas/numpy)
import numpy as np
import pandas as pd

tableau = np.array([1, 2, 3, 4, 5])
print(tableau * 2)      # [2 4 6 8 10]
````

## 9. Python Interactif — Le meilleur mode pour l'IA

### 9.1 Pourquoi le mode interactif ?

Quand vous travaillez sur des données ou de l'IA, vous avez besoin de :
- Tester un bout de code sans relancer tout le fichier
- Voir les résultats immédiatement
- Garder les variables en mémoire d'une cellule à l'autre
- Expérimenter rapidement

Le mode interactif combine le meilleur des deux mondes :
- **Organisation** d'un fichier `.py` (pas de Jupyter Notebook désorganisé)
- **Interactivité** de Jupyter (exécution cellule par cellule, résultats visuels)

### 9.2 Installer le kernel Jupyter

```bash
pip install ipykernel
```

### 9.3 Utiliser le mode interactif dans VS Code

- **Jupyter › Interactive Window › Text Editor: Execute Selection**

Dans un fichier `.py`, utilisez le raccourci **`Shift+Entrée`** :
- Sans sélection : exécute la ligne courante
- Avec une sélection : exécute le code sélectionné

Une fenêtre interactive s'ouvre à droite et affiche les résultats.

### 9.4 Délimiter des "cellules" dans un fichier `.py`

Vous pouvez structurer votre fichier en cellules avec `# %%` :

````python
# %% Cellule 1 : charger les données
import pandas as pd

data = {
    'nom': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'note': [15, 12, 18, 14],
    'ville': ['Amiens', 'Paris', 'Lyon', 'Amiens']
}
df = pd.DataFrame(data)
print(df)
````

````python
# %% Cellule 2 : analyser (df est toujours en mémoire)
print(f"Nombre d'étudiants : {len(df)}")
print(f"Note moyenne : {df['note'].mean():.1f}")
print(f"Meilleure note : {df['note'].max()}")
````

````python
# %% Cellule 3 : filtrer (toujours en mémoire)
amienois = df[df['ville'] == 'Amiens']
print("Étudiants d'Amiens :")
print(amienois)
````

### 9.5 Les Jupyter Notebooks (`.ipynb`)

Les notebooks Jupyter sont des fichiers `.ipynb` — c'est le format utilisé pour ce cours (les TDs).

**Différences avec un fichier `.py` interactif** :

| | Fichier `.py` + interactif | Notebook `.ipynb` |
|-|---------------------------|-------------------|
| Format | Fichier texte pur | Fichier JSON complexe |
| Git | Facile à versionner | Diffs difficiles à lire |
| Production | Utilisé en entreprise | Utilisé pour l'exploration |
| Usage recommandé | Projets, scripts | Cours, exploration, présentations |

Pour ce cours : les **TDs sont en `.ipynb`** (Colab), les **projets personnels seront en `.py`**.

## 10. Utiliser l'IA pour apprendre plus vite

Les assistants IA (ChatGPT, Claude) sont des outils très puissants pour apprendre Python. Voici comment les utiliser efficacement.

### Bons exemples de prompts

❌ **Mauvais** : *"Mon code ne marche pas, corrige-le"*

✅ **Bon** : *"J'apprends Python et j'essaie de calculer la moyenne d'une liste. Mon code : `sum(notes) / len(notes)`. Est-ce qu'il y a une façon plus efficace ?"*

### Stratégies efficaces

1. **Demander des explications** : "Explique-moi les fonctions Python simplement"
2. **Déboguer avec contexte** : coller le code ET le message d'erreur exact
3. **Demander des alternatives** : "Montre-moi 3 façons de résoudre ce problème"
4. **Vérifier la compréhension** : tester les suggestions et poser des questions de suivi

> L'IA est un outil pour **accélérer** l'apprentissage, pas pour le remplacer. L'objectif est de **comprendre**, pas d'avoir du code qui fonctionne sans savoir pourquoi.
